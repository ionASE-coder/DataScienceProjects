```python
from googleapiclient.discovery import build
import pandas as pd
```

```python
#importing video list
VIDEO_IDS=pd.read_excel("Youtube Links 2023 Sept. -2025 Sept. .xlsx").loc[:,"VideoID"].dropna().to_list()
VIDEO_IDS
```

```python
def get_youtube_comments(VIDEO_ID):
    # Initialize YouTube API client
    max_total_comments=200
    next_page_token = None
    collected=0

    try:
        video_response = youtube.videos().list(
            part='snippet',
            id=VIDEO_ID
        ).execute()

        if not video_response.get('items'):
            print(f"⚠️ Skipping {VIDEO_ID}: Video not found or inaccessible.")
            return  # safely skip this video
        
            
    except Exception as e:
        print(f"Error fetching comments for video {VIDEO_ID}: {e}")

    video_info = video_response['items'][0]['snippet']
    video_title = video_info['title']
    video_published_at = video_info['publishedAt']
    video_url = f"https://www.youtube.com/watch?v={VIDEO_ID}"

    while True:
        # Request a page of comments
        response = youtube.commentThreads().list(
            part='snippet',
            videoId=VIDEO_ID,
            maxResults=100,
            pageToken=next_page_token,
            textFormat='plainText'
        ).execute()

        # Extract comment details
        for item in response['items']:
            comment = item['snippet']['topLevelComment']['snippet']
            comments.append({
                'author': comment['authorDisplayName'],
                'comment': comment['textDisplay'],
                'likes': comment['likeCount'],
                'published_at': comment['publishedAt'],
                'video_title': video_title,
                'video_published_at': video_published_at,
                'video_url': video_url
            })
            collected += 1
            if collected >= max_total_comments:
                print(f"Reached {max_total_comments} comments for {VIDEO_ID}.")
                return

        # Check if there are more comments
        next_page_token = response.get('nextPageToken')
        if not next_page_token:
            break
```
