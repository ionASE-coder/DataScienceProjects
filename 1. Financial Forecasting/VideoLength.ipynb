```python
import yt_dlp
import pandas as pd

def get_video_length(video_id):
    url = f"https://www.youtube.com/watch?v={video_id}"
    
    ydl_opts = {
        "quiet": True,
        "skip_download": True,
    }
    
    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            info = ydl.extract_info(url, download=False)
            return info.get("duration", None)
    except Exception as e:
        print(f"Error for {video_id}: {e}")
        return None


def build_dataframe(video_ids):
    data = []
    
    for vid in video_ids:
        print(f"Processing {vid}...")
        duration = get_video_length(vid)
        
        data.append({
            "video_id": vid,
            "duration_sec": duration
        })
    
    df = pd.DataFrame(data)
    return df


# 🔹 Example usage
video_ids = pd.read_excel("Youtube Links 2023 Sept. -2025 Sept. .xlsx").loc[:,"VideoID"].dropna().to_list()

df = build_dataframe(video_ids)

print(df)
```

```python
df["duration_min"]=df['duration_sec']/60
df["duration_min"].describe()
cum_sum=df['duration_min'].sum()
print(cum_sum)
```
